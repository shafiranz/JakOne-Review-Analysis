```python
from google_play_scraper import reviews_all
import pandas as pd
```

```python
app_id = "com.dev.jakone.mbanking"
```

```python
reviews = reviews_all(
    app_id,
    lang='id',        
    country='id'      
)
```

```python
df = pd.DataFrame(reviews)
df = df[['content', 'score']]
df.rename(columns={'content': 'text', 'score': 'rating'}, inplace=True)
```

```python
def get_kategori(text):
    text = str(text).lower()
    if any(kw in text for kw in ['error', 'gagal', 'tidak bisa', 'bug', 'crash']):
        return 'Bug Teknis'
    elif any(kw in text for kw in ['tampilan', 'desain', 'interface', 'bagus']):
        return 'UI/UX'
    elif any(kw in text for kw in ['fitur', 'fungsi', 'update']):
        return 'Fitur & Fungsi'
    elif any(kw in text for kw in ['transaksi', 'top up', 'saldo', 'transfer']):
        return 'Masalah Transaksi'
    else:
        return 'Lainnya'
df['kategori'] = df['text'].apply(get_kategori)
```

```python
df.to_csv("kode_scraping.csv", index=False, encoding='utf-8')

print("Jumlah data yang berhasil diambil:", len(df))
df.head()
```
